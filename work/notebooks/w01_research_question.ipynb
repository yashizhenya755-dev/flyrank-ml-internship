## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision: which pages go into a content reviewer's limited weekly review
queue, and in what order.

Who acts: a content reviewer or strategist with limited time — they open the
top-ranked pages and decide to refresh, expand, protect, prune, or monitor
each one.

Cost of a wrong call:
- False positive (flagged, but didn't need review): wastes reviewer time
  that could have gone to a page that actually needed it.
- False negative (needed review, not flagged): a declining page keeps
  losing visibility/traffic unnoticed until the next review cycle.

Because review time is limited, this is a ranking problem (get the top of
the list right) more than a plain accuracy problem.

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

This work can say: which pages show signs of decline or under-performance
based on observed search and content signals, and can rank them for a
reviewer's limited time — a decision-support tool, not an automated fix.

This work cannot say: that a refresh causes recovery (no experiment here,
so no causal proof), that it has found a Google ranking factor, or that
the current trend_direction label is a reliable long-term target — it's a
same-window proxy rather than a future outcome. A stronger version of this
project would use a future-window label (prior 90 days -> next 30 days)
once working with the full warehouse release.

This work will not use FlyRank's own product scores (health_score,
priority_score, action_type) as model inputs, since that would just teach
a model to copy an existing decision instead of finding real signal.
